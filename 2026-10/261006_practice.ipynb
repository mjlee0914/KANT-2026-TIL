```typescript
// 5장 2강 문제 2-1 쿠폰 종류에 따른 혜택 안내 함수 작성하기
type TextCoupon = {
    name: string;
    code: string;
}

type PercentCoupon = {
    name: string;
    discountPercent: number;
}

function applyCoupon(coupon: TextCoupon | PercentCoupon) {
    if ('code' in coupon) {
        console.log("apply code coupon:", coupon.code)
    }
    if ('discountPercent' in coupon) {
        console.log("apply discount coupon:", coupon.discountPercent)
    }
}

const tc: TextCoupon = {
    name: "txtcp1",
    code: "EVENTCOUPON12345"
}

const pc: PercentCoupon = {
    name: "evtcp1",
    discountPercent: 50
}

applyCoupon(tc)
applyCoupon(pc)
```

```typescript
// 5장 2강 문제 3-1 서로소 판별 유니온과 switch 문을 활용한 주문 상태 처리
type PendingOrder = {
    status: "pending"
}

type ShippedOrder = {
    status: "shipped"
}

type DeliveredOrder = {
    status: "delivered"
}

type OrderState = PendingOrder | ShippedOrder | DeliveredOrder;

function processOrder(state: OrderState) {
    switch (state.status) {
        case "pending":
            console.log("주문 대기 안내")
            break;

        case "shipped":
            console.log("운송장 번호 출력")
            break;

        case "delivered":
            console.log("배송 완료 날짜")
            break;
    }
}

const stateP: PendingOrder = {status: "pending"};
const stateS: ShippedOrder = {status: "shipped"};
const stateD: DeliveredOrder = {status: "delivered"};

processOrder(stateP);
processOrder(stateS);
processOrder(stateD);
```

```typescript
// 타입 지정한 변수 호출 안하고 더 간단하게 함수 호출하는 방법
processOrder({ status: "pending" });
processOrder({ status: "shipped" });
processOrder({ status: "delivered" });
```

```typescript
// 6장 1강 문제 1-1 제네릭 함수를 활용한 배열 요소 추출기 구현
function getLastElement<T>(val: T[]) {
    if (val.length === 0) {
        // will return undefined
        return; 
    }
    // will return the last element
    return val[val.length -1];
}

console.log(getLastElement([10, 20, 30]));
console.log(getLastElement(["사과", "바나나", "체리"]));
```

```typescript
// 6장 1강 문제 3-1 제네릭 함수와 인터페이스의 결합을 통한 응답 객체 생성
interface DataContainer<T> {
    timestamp: number;
    value: T;
}

function wrapData<T>(data: T){  

    // a container whose value must have type T
    const container: DataContainer<T> = {
        timestamp: Date.now(),
        value: data,
    }
    return container;
}

console.log(wrapData(100));
```

```typescript
// 6장 2강 문제 1-1 부가세 계산 generic function
interface HasPrice {
    price: number;
}

// HasPrice interface is conditioned on the generic function calculateTax
function calculateTax<T extends HasPrice>(item: T): number {
    return item.price * 0.1 ;
}

const tax = calculateTax({price: 1000000})
// const tax = calculateTax(100000)
console.log("tax:", tax)
```

```typescript
// 7장 1강 문제 1-1 도서 클래스 정의 및 가격 변경 메서드 구현
class Book {
    constructor(public title:string, private price: number) {}

    setPrice(newPrice: number) {
        if (newPrice < 0) {
            return;
        } 
        return this.price = newPrice;
    }

    printInfo() {
        return `Book Title: ${this.title} | Book Price: ${this.price}`;
    }
}

let hp = new Book("Harry Potter", 9000);
console.log(hp.printInfo());
hp.setPrice(10000);
console.log(hp.printInfo())
```

```typescript
// 7장 1강 문제 2-1 상속을 통한 커피 클래스 구현
// parent class
class Beverage {
    constructor(public name: string, protected price: number) {}

    getDetails() {
        console.log(`Beverage name: ${this.name} | Price: ${this.price}`);
    }
}
class Coffee extends Beverage {
    // add property
    private shotCount: number;


    constructor(name: string, price: number, shotCount: number) {
        super(name, price);
        this.shotCount = shotCount;
    }

    printCoffeeInfo() {
        console.log(`shot count: ${this.shotCount}`);
    }
}
const latte = new Coffee("Cafe Latte", 5000, 2);
latte.printCoffeeInfo();
```

```typescript
// 7장 1강 문제 3-1 VIP 회원의 포인트 적립 로직 오버라이딩

// parent class
class User {
    constructor(public userName: string, protected point: number) {}

    addPoint(amount: number) {
        return this.point += amount;
    }
}

class VIPUser extends User {
    // add property
    private bonusRate: number;

    constructor(userName: string, point: number, bonusRate: number) {
        super(userName, point);
        this.bonusRate = bonusRate;
    }

    addPoint(amount: number): number {
        let bonusNum = amount * this.bonusRate;
        return this.point += bonusNum;
    }
}

let userEx = new User("lee", 50);
userEx.addPoint(20);
console.log(userEx);

let vipEx = new VIPUser("kim", 50, 30);
vipEx.addPoint(20); // bounsNum = 600
console.log(vipEx);
```

```typescript
// 7장 2강 문제 1-1 추상 클래스를 활용한 동물 클래스 정의
abstract class Animal {
    constructor(public name: string) {}

    printName() {
        console.log(this.name);
    }

    abstract makeSound(): void;
}

class Dog extends Animal {
    makeSound(): void {
        console.log("멍멍!!!");
    }
}

const john = new Dog("johnny");
john.printName();
john.makeSound();
```

```typescript
// 7장 2강 문제 2-1 인터페이스 구현과 다형성을 활용한 전자기기 제어

interface Powerable {
    turnOn(): void;
}

class Television implements Powerable {
    turnOn() {
        console.log("TV 전원이 켜졌습니다.");
    }
}

class AirConditioner implements Powerable {
    turnOn() {
        console.log("AC 전원이 켜졌습니다.");
    }
}

// Powerable 타입의 배열을 생성하고 Television 객체와 AirConditioner 객체를 저장
// type should be arr
// class can be inside an array if written as an instance
const arr: Powerable[] = [new Television(), new AirConditioner()];

// for...of iterator
for (const i of arr) {
    i.turnOn();
}
```
