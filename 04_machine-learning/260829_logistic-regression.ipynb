## 2. Key Concepts

#### Linear Regression vs. Logistic Regression (difference, process, equations, etc.)

[Both]
- supervised learning
- to predict the value of an output variable from one or multiple input variables
- output var X = dependent var
- input var y = independent var

[Linear Regression]
- formula:
$$
\hat{y} = \beta_0 + \beta_1 x
$$

- for multiple features:
$$
\hat{y} = \beta_0 + \beta_1x_1 + \beta_2x_2 + \cdots + \beta_p x_p
$$
  
[Logistic Regression]
- formula:
$$
z = \beta_0 + \beta_1x_1 + \cdots + \beta_p x_p
$$

| **Aspect** | **Linear Regression** | **Logistic Regression** |
|---|---|---|
| **What is it?** | Predicts a continuous output value from input variables. | Predicts the probability of an outcome belonging to a category. |
| **Relationship** | Linear relationship; represented by a straight line. | Logistic/sigmoidal relationship; represented by an S-shaped curve. |
| **Equation** | $\hat{y} = \beta_0 + \beta_1x$ | $p = \frac{1}{1 + e^{-z}}$ |
| **Type of supervised learning** | Regression | Classification |
| **Distribution type** | Normal (Gaussian) | Binomial |
| **Output** | Continuous numerical value | Probability between 0 and 1 |
| **Best suited for** | Predicting continuous values, such as price, salary, or temperature. | Predicting categorical outcomes, such as yes/no or spam/not spam. |


#### Classfication
데이터를 어떤 그룹에 넣을지 결정

Binary Classification
- classes: 0 or 1
- action function : Sigmoid
- Loss function : BCE

Multiclass Classification
- 여러개 클래스 중 하나 선택 ex) 0= dog, 1= cat, 3= bird
- action function: Soft-Max
- Loss Function: CCE, SCCE, Embedding Layer

#### Action Function - Sigmoid, Soft-Max

Sigmoid
- Binary Classification
- z = 0 ~ 1, 여러 숫자를 0~1의 사이의 확률로 바꿔주는 함수
- ex) Sigmoid output=0.9 -> 90% proabability, class 1

Soft-Max
- Multiclass Classification
- 여러 후보 중 각각의 확률을 계산하는 함수
- 여러개 class 점수를 확률로 바꿈
- 결과의 모든 확률 합 = 1
- ex) cat=0.1, dog=0.7, bird=0.2 total=1

#### Cross Entropy(Loss Function) - BCE, CCE, SCCE
모델의 예측이 정답과 얼마나 다른지 계산함, loss 작을수록 예측 잘함 판단

BCE Binary Class Entropy
- for 2 classes

Categorical Class Entoropy
- for multiclass + one-hot encoding ex) [0, 0, 1]
- 두 배열의 모든 원소를 매칭하여 연산 -> 클래스 많아질수록 메모리 낭비

SCCE Sparse Categorical Class Entropy
- for multiclass + int label ex) 2
- 정답 인덱스 숫자 하나로 저장함 -> 메모리 효율 (One-Hot Encoding과 비교)
```
Loss Function
│
├── MSE (Mean Squared Error)
├── MAE (Mean Absolute Error)
├── Cross Entropy
│   ├── BCE (Binary Cross Entropy)
│   ├── CCE (Categorical Cross Entropy)
│   └── SCCE (Sparse Categorical Cross Entropy)
└── 기타 Loss Functions
```

#### K-Fold Cross Validation
- 전체 데이터를 K개로 나누어 데이터 선택 편향을 줄임 overfitting X
- 나누기 전 shuffle -> 각 데이터가 공평하게 들어가도록 함 ex) randome_state=42

#### One-Hot Encoding
- 범주형 데이터 전처리 방식
- 1, 2, 3으로 단순하게 데이터가 나뉘면 모델이 가중치를 '숫자에 따른 서열'로 인식하여 편향적인 계산을 하게 됨 
- 모든 범주에게 공평하게 독립된 열을 분양하고, 해당 품종 열=1, 그렇지 않은 열=0으로 채워 가중치를 동등하게 다루도록 유도함
- 클래스가 많을수록 메모리 소비가 심해질 수 있음 "차원의 저주"
- 다중공선성(dummy variable trap): 2개 열->하나가 0이면 나머지는 무조건 1 -> 하나를 삭제 ex) 평일=1, 주말=0 두개의 열 -> 주말 열 삭제 -> 평일=0, 1만 남게 함 `drop='first'`

#### Embedding Layer
- 범주가 많을 때 사용
- one-hot encocoding의 경우 데이터 십만 개 -> 십만 차원 분양 -> 메모리 낭비
- embedding에서 작은 숫자 벡터로 바꿔 십만 차원 -> 128차원으로 줄임

## 3. Practice 실습 코드

````python
# chap03-2 문제1
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# 데이터 불러오기
df = pd.read_csv("Iris.csv")

# 원본 파일에 영향 없도록 복사본 데이터 생성
bin_df = df[df['Species'].isin(['Iris-versicolor', 'Iris-virginica'])].copy()

# 입력변수 = SepalLengthCm, SepalWidthCm 두개의 feature 꺼내기
bin_X= bin_df[['SepalLengthCm', 'SepalWidthCm']]

# 정답타깃변수 = Iris-versicolor:0, Iris-virginica: 1 설정
bin_y=  bin_df['Species'].map({'Iris-versicolor':0, 'Iris-virginica':1}) 

# data split: train 80%, test 20%, random 42, stratify bin_y
train_bin_X, test_bin_X, train_bin_y, test_bin_y = train_test_split(
    bin_X, bin_y, test_size=0.2, random_state=42, stratify=bin_y
)

# logistic regression model object
binary_logi_model = LogisticRegression() 

# 모델학습 수행
binary_logi_model.fit(train_bin_X, train_bin_y)

# 모델테스트 수행
bin_test_prediction = binary_logi_model.predict(test_bin_X)

# 최종 테스트 정확도 계산 및 출력
bin_accuracy = accuracy_score(test_bin_y, bin_test_prediction)
print("최종 테스트 정확도: %.4f" % bin_accuracy)
````
