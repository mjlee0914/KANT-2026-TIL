```python
class user_service:
    def __init__(self):
        data = supabase.auth.get_user()
        if data: self.user = data.user

    def create(self, cellphone, zipcode, address, address_sub, type = "BUYER"):
        response = (
            supabase.table('user_details')
                .insert({
                    "user_id" : self.user.id,
                    "type" : type,
                    "cellphone" : cellphone,
                    "zipcode" : zipcode,
                    "address" : address,
                    "address_sub" : address_sub
                })
                .execute()
        )
        return response.data
    
    def update(self, cellphone=None, zipcode=None, address=None, address_sub=None, type = None):
        update_data = {}
        if cellphone is not None:
            update_data["cellphone"] = cellphone
        if zipcode is not None:
            update_data["zipcode"] = zipcode
        if address is not None:
            update_data["address"] = address
        if address_sub is not None:
            update_data["address_sub"] = address_sub
        if type is not None:
            update_data["type"] = cellphone
        
        if not update_data:
            return None
        
        response = (
            supabase.table('user_details')
                .update(update_data)
                .eq("user_id", self.user.id)
                .execute()
        )
        return response.data

    def get(self):
        response = (
            supabase.table("user_details")
                .select("*")
                .eq("user_id",self.user.id)
                .execute()
        )
        return response.data
```
