### 2.2 매개변수 Parameter and 인수 Argument
- Parameter: 함수의 '선언부' 공간에 위치
- Argument: 실제 함수 작동 시 해당 통로를 통해 전달하는 '실제 데이터', 함수 '호출부' 공간에서 주입

````python
def taste(food): # food = parameter
    print(f"{food}는(은) 맛있다")

taste("떡볶이") # arguments
taste("사과")
````

### 2.3 위치 인수 Positional Arguments and 키워드 인수 Keyword Arguements
위치 인수
- parameter 순서대로 인수를 차례대로 주입
- 순서 바뀌면 오작동 유발 가능 주의

키워드 인수
- `매개변수명=값` 형태
- 순서 상관 X
- 위치 인수와 작성시 반드시 위치 인수 먼저!

````python
def calc(num1, num2):
    return num1+num2

print(calc(1, 2)) # 위치 인수   
print(calc(num2=20, num1=10)) # 키워드 인수
````

### 2.4 Default Parameter 기본값 매개변수
- 함수 정의 시 매개변수에 특정 값을 미리 대입 
- 특정 값이 매겨진 매개변수는 함수 호출 시 해당 인수 생략 가능
- 기본값이 없는 매개변수보다 뒤에 위치해야 함 (interpreter가 인수를 앞에서부터 매핑함)

````python
def introduce(name, age=20):
    print(f"이름은 {name}이고, 나이는 {age}세입니다.")

introduce("철수")       # 나이 인수를 생략하면 기본값 20이 자동 적용
introduce("지영", 25)   # 인수를 전달하면 25로 갱신되어 작동
````

### 2.5 가변 인수 *args, 키워드 가변 인수 **kwargs
- `*args`
  - 입력될 인수가 몇 개인지 아직 정해지지 않았을 때 매개변수에 적용
  - 출력시 하나의 **튜플 구조**로 묶여 차례대로 순회가 가능

- -`*kwargs`
  - 인수 이름을 지정하여 '키-값' 형태로 전달
  - 입력된 모든 쌍 데이터를 하나의 **딕셔너리 구조**로 변환하여 처리

````python
def print_user_info(name, age, *args, **kwargs):
    print("name:", name)
    print("age:",  age)
    print(args)
    print(kwargs)

print_user_info("김철수", 20, 'A', 20, mobie="010-0000-0000") # 'A', 20은 Tuple로 *args에 들어감
````

### 2.6 return
- 함수 내부에서 연산된 최종 결과값을 내보내는 통로 역할
- interpreter가 return 만나면 즉시 함수 종료&탈출
- **복수 데이터 반환** 가능: **튜플** 구조로 묶임, unpacking 가능

````python
def calc(num1, num2):
    return num1 + num2, num1 * num2, num1 - num2, num1 / num2 # return 값 여러개면 튜플 하나로 반환

print(calc(2,2))
x = a, b, c, d = calc(2,2) # unpacking
print(x)
````

````python
# 문제 1-1: 대여 완료 안내 메시지 출력
def rent_book(user_name, book_title):
    print(f"[회원 이름: {user_name} | 책 제목: {book_title}] 대여 완료")

rent_book("홍길동", "파이썬 안내서")
rent_book(book_title="데이터 구조", user_name="이영희")
````

````python
# 문제 2-1: 배송비 자동 계산 및 다중 데이터 반환
def calculate_total(price, delivery_fee = 3000):
    return price + delivery_fee, "결제가 완료되었습니다"

calculate_total(25000)
total, message = calculate_total(25000)
print(total, message)
````

````python
# 문제 3-1: 무제한 과목 점수 합계 및 평균 산출
def get_total_and_average(*arg):
    if len(arg) == 0:
        return 0
    else: 
        total = sum(arg)
        avg = total / len(arg)
        return total, avg
       

print(get_total_and_average(90, 85, 95)) 
print(get_total_and_average(70, 80, 90, 100, 65))
````
