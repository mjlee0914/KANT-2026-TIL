```python
# chap7-01 문제 2
bunsik = ["떡볶이", "김밥", "어묵"]

try:
    n = int(input("조회할 메뉴의 인덱스 번호를 입력하세요. (0, 1, 2)"))
    result = bunsik[n]


except ValueError:
    print("숫자만 입력 가능합니다.")

except IndexError:
    print("존재하지 않는 위치의 데이터")

else: 
    print("선택한 메뉴:", result)

finally:
    print("프로그램을 종료합니다")
```

```python
# chap7-01 문제 3
def order(order_num):
    if order_num <= 0:
        raise ValueError("1 이상의 주문 수량을 입력해 주세요.")

    print("주문 수량이 정상적으로 설정되었습니다.")

try:
    minus_order = -1
    order(minus_order)

except ValueError:
    print("[경고] 주문 수량은 -1이 될 수 없습니다.")
```

```python
# chap7-01 문제3 variation
class NoMinusZero(Exception):
    def __str__(self):
        return "1 이상의 주문 수량을 입력해 주세요."
```

```python
def order(order_num):
    if order_num <= 0:
        raise NoMinusZero
    print("주문 수량이 정상적으로 설정되었습니다.")
```

```python
try:
    order(-1)
except NoMinusZero as e:
    print(e)
```

```python
# chap7-02 문제1
def calculate_share(total_price, num_people):
    share_per_person = total_price / num_people
    return share_per_person

def start_settlement():
    participants = 0
    result = calculate_share(50000, participants)
    print(result)

start_settlement()
```

```python
# chap7-02 문제2
def calculate_revenue():
    total_revenue = 0
    daily_sales = [12000, 25000, 31000]
    for amount in daily_sales:
        total_revenue += amount
    print(total_revenue)

calculate_revenue()
```

```python
# chap7-02 문제3
def sum_scores():
    final_score = 50
    bonus_scores = [10, 20, 30]
    for score in bonus_scores:
        final_score += score
    return final_score

result = sum_scores()
```
