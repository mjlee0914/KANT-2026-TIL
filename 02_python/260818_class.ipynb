```python
# chap8-1 문제 1
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def info(self):
        print(f"책 제목: {self.title}, 저자: {self.author}")


book1 = Book("싯다르타", "헤르만 헤세")
book2 = Book("오뒷세이아", "호메로스")

book1.info()
book2.info()
```

```python
# chap8-1 문제 2
class SmartLamp:
    def __init__(self, model):
        self.model = model
        self.is_on = False

    def on(self):
        self.is_on = True
        print(f"{self.model} 조명이 켜졌습니다.")

    def off(self):
        self.is_on = False
        print(f"{self.model} 조명이 꺼졌습니다.")


smart_lamp = SmartLamp("M05")
smart_lamp.on()
smart_lamp.off()
```

```python
# chap8-1 문제 3
class Hero:
    def __init__(self, name):
        self.name = name
        self.hp = 100

    def take_damage(self, damage):
        self.hp -= damage
        print(self.hp)

        if self.hp <= 0:
            print(f"{self.name} 캐릭터가 쓰러졌습니다.")

cha = Hero("Spiderman")
cha.take_damage(60)
cha.take_damage(20)
cha.take_damage(10)
cha.take_damage(10)
```

```python
# chap8-2 문제 1
class Phone:
    def call(self):
        print("통화를 시작합니다")

class SmartPhone(Phone):
    def call(self):
        print("영상 통화를 시작합니다")

    def search_web(self):
        print("인터넷 검색을 시작합니다")


iPhone = SmartPhone()
iPhone.call()
iPhone.search_web()
```
